## **3. Contenido de la Clase**  
### **Situación Real ➔ Modelo Matemático**  
**Problema 1 (Pensamiento Numérico):**  
- **Contexto:** En el mercado de Caucasia, los niños ven montones de plátanos.  
- **Pregunta:** *"¿Cómo sabemos cuántos plátanos hay sin contarlos uno por uno?"*  
- **Modelo:**  
  - **Cantidad:** Montón de 5 plátanos → Percepción visual.  
  - **Número:** Símbolo "5" → Abstracción.  
  - **Código Python:**  
    ```python  
    import matplotlib.pyplot as plt  
    plt.bar(['Cantidad', 'Número'], [5, 5], color=['orange', 'blue'])  
    plt.title('Relación entre 5 plátanos y el número "5"')  
    plt.show()  
    ```  

**Problema 2 (Pensamiento Métrico):**  
- **Contexto:** Medir la longitud de una hamaca con palmos.  
- **Modelo:**  
  - **Cantidad:** 6 palmos → Magnitud física.  
  - **Número:** "6" → Representación simbólica.

## **4. Mapa Conceptual (Graphviz)**  
```python  
from graphviz import Digraph  

graph = Digraph(engine='dot', format='png', graph_attr={  
    'overlap': 'scalexy',  # Evita superposiciones  
    'bgcolor': 'lightyellow'  
})  

# Nodos  
graph.node('A', 'Cantidad vs. Número', shape='box', style='filled', fillcolor='#FFD700')  
graph.node('B', 'Cantidad (Concreta)', shape='ellipse', color='blue')  
graph.node('C', 'Número (Abstracto)', shape='ellipse', color='red')  
graph.node('D', 'Ejemplo: 5 plátanos', shape='note', fillcolor='#FFA07A')  
graph.node('E', 'Símbolo "5"', shape='note', fillcolor='#98FB98')  

# Edges  
graph.edge('A', 'B', label='Percepción sensorial')  
graph.edge('A', 'C', label='Representación simbólica')  
graph.edge('B', 'D', label='Objetos reales')  
graph.edge('C', 'E', label='Código cultural')
```  
