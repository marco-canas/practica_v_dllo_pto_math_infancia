### 3. **Contenido de la clase**  
- **Situación real 1 (Pensamiento numérico):**  
  Un grupo de niños quiere repartir 12 chocolates entre 4 amigos. ¿Cuántos chocolates le tocan a cada uno?  
  - **Modelo matemático:**  
    Usando los bloques de Cuisenaire, se representa la división $12 \div 4 = 3$.  

- **Situación real 2 (Pensamiento algebraico):**  
  Un niño tiene 5 bloques rojos (cada uno representa el número 2) y quiere saber cuántos bloques blancos (número 1) necesita para igualar la longitud total.  
  - **Modelo matemático:**  
    Se plantea la ecuación $5 \times 2 = x \times 1$, donde $x = 10$.  

- **Situación real 3 (Pensamiento geométrico):**  
  Un estudiante quiere construir un rectángulo usando bloques de Cuisenaire de diferentes colores.  
  - **Modelo matemático:**  
    Se exploran las propiedades de los rectángulos y se calcula el área usando los bloques.

### 4. **Mapa conceptual**  
El mapa conceptual se genera con Graphviz y se renderiza en la ubicación especificada.  

```python
from graphviz import Digraph

# Crear el gráfico
graph = Digraph('Cuisenaire_Numbers', format='png')
graph.attr(overlap='false')  # Evitar superposiciones

# Nodos
graph.node('A', 'Números en Color de Cuisenaire', shape='ellipse', color='blue', style='filled', fillcolor='lightblue')
graph.node('B', 'Pensamiento Numérico', shape='box', color='red')
graph.node('C', 'Pensamiento Algebraico', shape='box', color='green')
graph.node('D', 'Pensamiento Geométrico', shape='box', color='orange')
graph.node('E', 'Aplicaciones Artísticas', shape='box', color='purple')

# Edges
graph.edge('A', 'B', label='Operaciones básicas')
graph.edge('A', 'C', label='Ecuaciones simples')
graph.edge('A', 'D', label='Formas y áreas')
graph.edge('A', 'E', label='Diseños y patrones')

# Renderizar y mostrar
graph.render("C:/Users/marco/Descargas/Cuisenaire_Numbers", view=True)
```
