## promp para convertir el archivo .docx en .md   

Dame una función de usuario de python que tranforme un archivo .docx en un archivo .md. dado como parámetro al archivo_path  y que guarde el archivo .md en la misma carpeta donde está el archivo .docx  





Aquí tienes una función en Python que convierte un archivo `.docx` en un archivo `.md` usando la biblioteca `python-docx` para leer el archivo de Word y `markdownify` para convertir el contenido a Markdown. Asegúrate de instalar las dependencias antes de ejecutar el código:

Código de la función:  


````python
import os
import docx
from markdownify import markdownify as md

def docx_to_md(file_path):
    if not file_path.lower().endswith('.docx'):
        raise ValueError("El archivo debe tener la extensión .docx")

    folder_path = os.path.dirname(file_path)
    file_name = os.path.splitext(os.path.basename(file_path))[0]
    md_file_path = os.path.join(folder_path, f"{file_name}.md")

    doc = docx.Document(file_path)
    md_content = ""

    # Extraer párrafos
    for para in doc.paragraphs:
        md_content += md(para.text) + "\n\n"

    # Extraer tablas
    for table in doc.tables:
        table_md = []
        for row in table.rows:
            cols = [md(cell.text.strip() if cell.text else "") for cell in row.cells]
            table_md.append(" | ".join(cols))
        md_content += "\n" + "\n".join(table_md) + "\n\n"

    # Extraer imágenes
    for i, rel in enumerate(doc.part.rels):
        if "image" in doc.part.rels[rel].target_ref:
            image_data = doc.part.rels[rel].target_part.blob
            image_path = os.path.join(folder_path, f"{file_name}_image_{i+1}.png")

            with open(image_path, "wb") as img_file:
                img_file.write(image_data)

            md_content += f"\n![Imagen {i+1}]({os.path.basename(image_path)})\n\n"

    # Guardar el archivo en formato Markdown
    with open(md_file_path, "w", encoding="utf-8") as md_file:
        md_file.write(md_content)

    print(f"Archivo Markdown guardado en: {md_file_path}")
    return md_file_path

# Ejemplo de uso:
docx_to_md(r"C:\Users\marco\Documentos\docencia\practica_v_dllo_pto_math_infancia\clases_2025_1\documentos_actas\programa_practica_v_2025_1.docx")
````

Esta función:  
- Verifica que el archivo tenga la extensión `.docx`.  
- Lee el contenido del archivo de Word y lo convierte a Markdown.  
- Guarda el archivo `.md` en la misma carpeta del archivo original.

# Formato de asistencia al centro de práctica  

````python
from docx import Document
from docx.shared import Pt
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT

def crear_formato_seguimiento():
    # Crear un nuevo documento
    doc = Document()
    
    # Configuración básica (fuente y alineación)
    style = doc.styles['Normal']
    font = style.font
    font.name = 'Arial'
    font.size = Pt(11)
    
    # Título del documento
    titulo = doc.add_heading('FORMATO DE SEGUIMIENTO DE ASISTENCIA Y DESEMPEÑO', level=1)
    titulo.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Subtítulo
    subtitulo = doc.add_heading('Práctica Pedagógica V – Liceo Caucasia', level=2)
    subtitulo.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Datos del practicante
    doc.add_heading('Datos del Practicante:', level=3)
    doc.add_paragraph('Nombre: ___________________________________________________')
    doc.add_paragraph('Código/ID: ___________________')
    doc.add_paragraph('Semestre: ___________________')
    doc.add_paragraph('Fecha: ___________________')
    
    # Datos del grupo/grado
    doc.add_heading('Datos del Grupo/Grado Asignado:', level=3)
    doc.add_paragraph('Docente a cargo: ____________________________________________')
    doc.add_paragraph('Grado/Grupo: ___________________')
    doc.add_paragraph('Área/Asignatura: ___________________')
    
    # Registro de asistencia (tabla)
    doc.add_heading('Registro de Asistencia:', level=3)
    tabla_asistencia = doc.add_table(rows=2, cols=4)
    tabla_asistencia.style = 'Table Grid'
    
    # Encabezados de la tabla
    celdas_encabezado = tabla_asistencia.rows[0].cells
    celdas_encabezado[0].text = 'Hora de Llegada'
    celdas_encabezado[1].text = 'Hora de Salida'
    celdas_encabezado[2].text = 'Firma del Practicante'
    celdas_encabezado[3].text = 'Firma del Docente a Cargo'
    
    # Líneas en blanco para llenar
    celdas_datos = tabla_asistencia.rows[1].cells
    for celda in celdas_datos:
        celda.text = ' '  # Espacio para escribir
    
    # Evaluación de desempeño
    doc.add_heading('Evaluación de Desempeño:', level=3)
    doc.add_paragraph('(Marque con ✔ según corresponda y agregue comentarios)')
    
    # Lista de criterios de evaluación
    criterios = [
        "1. Puntualidad: ☐ Excelente ☐ Buena ☐ Regular ☐ Deficiente",
        "2. Participación en Clase: ☐ Muy activo/a ☐ Participativo/a ☐ Poco participativo/a ☐ Pasivo/a",
        "3. Relación con Estudiantes: ☐ Excelente ☐ Buena ☐ Regular ☐ Requiere mejora",
        "4. Dominio de Contenidos: ☐ Alto ☐ Adecuado ☐ Básico ☐ Insuficiente",
        "5. Creatividad y Metodología: ☐ Innovador/a ☐ Aplica estrategias efectivas ☐ En desarrollo ☐ Limitado/a",
    ]
    
    for criterio in criterios:
        doc.add_paragraph(criterio)
    
    # Recomendaciones
    doc.add_heading('Recomendaciones/Observaciones del Docente a Cargo:', level=3)
    for _ in range(3):  # 3 líneas para escribir
        doc.add_paragraph('___________________________________________________________')
    
    # Firma y fecha
    doc.add_paragraph('Firma del Docente: _________________________   Fecha: ___________')
    
    # Guardar el documento
    doc.save('Seguimiento_Practica_Pedagogica.docx')
    print("¡Documento generado con éxito! (Seguimiento_Practica_Pedagogica.docx)")

# Ejecutar la función
if __name__ == "__main__":
    crear_formato_seguimiento()
````
