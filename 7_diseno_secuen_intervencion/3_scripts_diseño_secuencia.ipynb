# Script para convertir el Ejemplo de secuencia didáctica de .md a .docx 

````python
from docx import Document
from docx.shared import Pt
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT

# Crear un nuevo documento Word
doc = Document()

# Añadir título principal
title = doc.add_heading('EJEMPLO DE ESTRUCTURA REORGANIZADA PARA LA SECUENCIA DE ENSEÑANZA', level=1)
title.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER

# Añadir subtítulo
subtitle = doc.add_heading('SECUENCIA DIDÁCTICA: "MATEMÁTICAS EN NUESTRO TERRITORIO"', level=2)
subtitle.alignment = WD_PARAGRAPH_ALIGNMENT.CENTER

# Añadir información de grado y duración
doc.add_paragraph('Grado: Preescolar a 3° de primaria')
doc.add_paragraph('Duración: 7 sesiones (ajustar según necesidad)')

# Añadir línea divisoria
doc.add_paragraph('---')

# Sección 1: Descripción del problema educativo
doc.add_heading('1. DESCRIPCIÓN DEL PROBLEMA EDUCATIVO', level=2)
doc.add_paragraph('(Basado en los DBA del MEN Colombia y la caracterización realizada)')
doc.add_paragraph('- Problema central: Dificultades en el desarrollo del pensamiento numérico (conteo, clasificación y representación simbólica de cantidades), evidenciado en la caracterización mediante:')
doc.add_paragraph('  - Bajo dominio de correspondencia uno a uno en preescolar (DBA: "Establece relaciones de cantidad").')
doc.add_paragraph('  - Confusión entre registros gráficos y simbólicos en 1°-3° (DBA: "Usa diferentes representaciones para expresar cantidades").')
doc.add_paragraph('  - Escasa vinculación de las matemáticas con contextos cotidianos (DBA: "Resuelve problemas relacionados con su entorno").')
doc.add_paragraph('- Brecha identificada: Los niños no superan el nivel icónico (Dienes) y presentan resistencia a la abstracción, según los resultados de la caracterización.')

# Sección 2: Justificación de los enfoques pedagógicos
doc.add_heading('2. JUSTIFICACIÓN DE LOS ENFOQUES PEDAGÓGICOS', level=2)
doc.add_paragraph('(Inferidos de las necesidades educativas detectadas)')
doc.add_paragraph('1. Aprendizaje Basado en Juegos (ABJ):')
doc.add_paragraph('   - Fundamentación: Permite transitar de lo concreto a lo abstracto mediante manipulación de materiales (Piaget).')
doc.add_paragraph('   - Vinculación con la necesidad: Los niños requieren reforzar el conteo y la clasificación en contextos lúdicos, dado su bajo desempeño en actividades simbólicas puras.')
doc.add_paragraph('2. Enfoque Sociocultural:')
doc.add_paragraph('   - Fundamentación: La mediación cultural (Vygotsky) facilita aprendizajes significativos.')
doc.add_paragraph('   - Vinculación con la necesidad: La caracterización mostró desconexión entre los contenidos matemáticos y las prácticas locales (ej: comercio de frutas).')
doc.add_paragraph('3. Registros de Representación Semiótica (Duval):')
doc.add_paragraph('   - Fundamentación: La coordinación entre registros (concreto-gráfico-simbólico) es clave para la comprensión.')
doc.add_paragraph('   - Vinculación con la necesidad: Los errores en evaluaciones previas revelaron dificultades para traducir entre representaciones.')

# Sección 3: Objetivos de aprendizaje por clase
doc.add_heading('3. OBJETIVOS DE APRENDIZAJE POR CLASE', level=2)
doc.add_paragraph('(7 sesiones alineadas a los DBA y enfoques)')
doc.add_heading('Sesión 1: "Contamos lo que sembramos"', level=3)
doc.add_paragraph('- Objetivo: Desarrollar correspondencia uno a uno y conteo hasta 20 usando semillas autóctonas (DBA: "Cuenta colecciones de objetos").')
doc.add_heading('Sesión 2: "Clasificamos nuestra cosecha"', level=3)
doc.add_paragraph('- Objetivo: Identificar atributos para clasificar objetos (color/tamaño) vinculados a cultivos locales (DBA: "Reconoce propiedades de objetos").')
doc.add_heading('Sesión 3: "Del campo al mercado"', level=3)
doc.add_paragraph('- Objetivo: Representar cantidades en registros concretos (frutas), gráficos (tablas) y simbólicos (números) (DBA: "Usa múltiples representaciones").')
doc.add_heading('Sesión 4: "Sumamos nuestras ganancias"', level=3)
doc.add_paragraph('- Objetivo: Resolver problemas aditivos simples con monedas ficticias (DBA: "Soluciona situaciones cotidianas con operaciones básicas").')
doc.add_heading('Sesión 5: "Patrones en nuestra artesanía"', level=3)
doc.add_paragraph('- Objetivo: Identificar y crear patrones geométricos con materiales locales (DBA: "Reconoce regularidades en secuencias").')
doc.add_heading('Sesión 6: "Medimos la tierra"', level=3)
doc.add_paragraph('- Objetivo: Comparar longitudes usando unidades no convencionales (ej: hojas de plátano) (DBA: "Establece relaciones de medida").')
doc.add_heading('Sesión 7: "Evaluamos lo aprendido"', level=3)
doc.add_paragraph('- Objetivo: Integrar habilidades mediante la resolución de un problema contextualizado (ej: "Organizar una feria escolar").')

# Sección de sesiones de intervención
doc.add_heading('SESIONES DE INTERVENCIÓN', level=2)
doc.add_paragraph('(Basadas en los DBA del MEN Colombia y los enfoques seleccionados)')
doc.add_paragraph('---')

# Sesión 1
doc.add_heading('Sesión 1: "Contamos lo que sembramos"', level=3)
doc.add_paragraph('Enfoque principal: Aprendizaje Basado en Juegos (ABJ) + Enfoque Sociocultural')
doc.add_paragraph('Objetivo: Desarrollar correspondencia uno a uno y conteo hasta 20 usando semillas autóctonas (DBA: "Cuenta colecciones de objetos").')
doc.add_paragraph('Actividades:')
doc.add_paragraph('1. Juego de la cosecha:')
doc.add_paragraph('   - Materiales: Semillas de maíz/plátano, canastos pequeños, carteles con números (1-20).')
doc.add_paragraph('   - Desarrollo:')
doc.add_paragraph('     - Los niños simulan recolectar semillas en canastos (1 semilla = 1 número dicho en voz alta).')
doc.add_paragraph('     - Validan conteo comparando con carteles numéricos.')
doc.add_paragraph('   - Evaluación: Lista de cotejo para verificar correspondencia y secuencia correcta.')
doc.add_paragraph('2. Canción numérica:')
doc.add_paragraph('   - Usan una canción tradicional local (ej: "Los pollitos") para contar hacia adelante y atrás.')

# Continuar con las demás sesiones (similar a la sesión 1)
# ...

# Sección de recursos por sesión
doc.add_heading('RECURSOS POR SESIÓN', level=2)
table = doc.add_table(rows=1, cols=3)
table.style = 'Table Grid'
hdr_cells = table.rows[0].cells
hdr_cells[0].text = 'Sesión'
hdr_cells[1].text = 'Materiales clave'
hdr_cells[2].text = 'Adaptación contextual'
row_cells = table.add_row().cells
row_cells[0].text = '1'
row_cells[1].text = 'Semillas, canastos'
row_cells[2].text = 'Uso de cultivos locales (maíz/plátano)'
row_cells = table.add_row().cells
row_cells[0].text = '3'
row_cells[1].text = 'Frutas, billetes de papel'
row_cells[2].text = 'Moneda local ficticia ("pesos caucasia")'
row_cells = table.add_row().cells
row_cells[0].text = '5'
row_cells[1].text = 'Hojas de plátano, arcilla'
row_cells[2].text = 'Motivos artesanales veredales'

# Nota final
doc.add_heading('NOTA:', level=2)
doc.add_paragraph('Cada sesión retoma los tres enfoques de manera integrada, pero se enfatiza uno según el objetivo. Por ejemplo:')
doc.add_paragraph('- ABJ domina en sesiones lúdicas (1, 4, 5).')
doc.add_paragraph('- Sociocultural es clave en actividades de mercado (3, 6).')
doc.add_paragraph('- Registros Semióticos se trabaja explícitamente en traducciones (2, 3, 5).')

# Guardar el documento
doc.save('4_ejemplo_de_secuencia_didactica.docx')

print("Documento 'Ejemplo_de_secuencia_didactica.docx' creado exitosamente.")
````

# Script para convertir el Taller de diseño de secuencias didácticas de .md a .docx 

````python
from docx import Document
from docx.shared import Pt, RGBColor
from docx.enum.text import WD_PARAGRAPH_ALIGNMENT, WD_BREAK
from docx.oxml.ns import qn
from docx.oxml import OxmlElement

def create_workshop_document():
    # Crear documento
    doc = Document()
    
    # Configuración de estilos
    style = doc.styles['Normal']
    font = style.font
    font.name = 'Calibri'
    font.size = Pt(11)
    
    # Título introductorio
    intro = doc.add_paragraph()
    intro.add_run('Este es un ').bold = False
    intro.add_run('diseño lúdico y participativo').bold = True
    intro.add_run(''' para que los maestros en formación aprendan a crear secuencias didácticas 
                  basadas en la caracterización y los enfoques pedagógicos, en un taller de ''')
    intro.add_run('2 horas').bold = True
    intro.add_run('. Incluye dinamismo, trabajo colaborativo y conexión con el contexto de Caucasia:')
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Título principal del taller
    title = doc.add_heading('TALLER: "DISEÑANDO SECUENCIAS CON SABOR A CAUCASIA"', level=1)
    title_run = title.runs[0]
    title_run.bold = True
    title_run.font.color.rgb = RGBColor(0, 32, 96)  # Azul oscuro
    
    # Información básica
    info = doc.add_paragraph()
    info.add_run('Objetivo: ').bold = True
    info.add_run('Diseñar secuencias didácticas innovadoras para la Práctica V, integrando la caracterización realizada y 3-4 enfoques pedagógicos.\n')
    info.add_run('Duración: ').bold = True
    info.add_run('2 horas\n')
    info.add_run('Materiales:').bold = True
    
    # Lista de materiales
    materials = [
        "Pliegos de papel periodico, post-it, marcadores.",
        "Tarjetas con los 11 enfoques pedagógicos (impresos o digitales).",
        "Fotos/evidencias de la caracterización (ej: fotos de aulas, trabajos de niños).",
        "Cronómetro y música alegre (ej: música de carrilera para ambientar)."
    ]
    
    for material in materials:
        doc.add_paragraph(material, style='List Bullet')
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Dinámica 1
    dyn1 = doc.add_heading('Dinámica 1: "¡Aterrizando la caracterización!"', level=2)
    dyn1.add_run(' (30 min)').bold = False
    
    # Actividad
    act1 = doc.add_paragraph()
    act1.add_run('Actividad: ').bold = True
    act1.add_run('Galería de hallazgos').italic = True
    
    # Pasos
    steps = doc.add_paragraph()
    steps.add_run('1. Paso 1: ').bold = True
    steps.add_run('En grupos, los practicantes pegan en los pliegos de papel periódico:\n')
    steps.add_run('   - ').bold = True
    steps.add_run('3 necesidades prioritarias ').italic = True
    steps.add_run('identificadas en sus 8 horas de caracterización (ej: "Los niños confunden números con cantidades").\n')
    steps.add_run('   - ').bold = True
    steps.add_run('1 foto/evidencia ').italic = True
    steps.add_run('que respalde cada necesidad.\n')
    steps.add_run('2. Paso 2: ').bold = True
    steps.add_run('Rotación por estaciones. Cada grupo comenta sus hallazgos en 3 minutos.\n')
    steps.add_run('3. Producto: ').bold = True
    steps.add_run('Lista colaborativa de necesidades en el tablero.')
    
    # Frases clave
    frases = doc.add_paragraph()
    frases.add_run('Frases clave:').bold = True
    frases.add_run('\n"¿Qué nos dicen los niños con sus dificultades?"\n"¡Esto no lo vi en los libros, lo viví en el aula!"')
    frases.italic = True
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Dinámica 2
    dyn2 = doc.add_heading('Dinámica 2: "Enfoques pedagógicos en acción"', level=2)
    dyn2.add_run(' (40 min)').bold = False
    
    # Actividad
    act2 = doc.add_paragraph()
    act2.add_run('Actividad: ').bold = True
    act2.add_run('Ruleta de enfoques').italic = True
    
    # Pasos
    steps = doc.add_paragraph()
    steps.add_run('1. Paso 1: ').bold = True
    steps.add_run('Cada grupo elige ').bold = False
    steps.add_run('3 tarjetas al azar ').italic = True
    steps.add_run('(de los 11 enfoques estudiados). Ejemplo:\n')
    steps.add_run('   - Aprendizaje Basado en Juegos (ABJ).\n')
    steps.add_run('   - Enfoque sociocultural.\n')
    steps.add_run('   - Registros semióticos (Duval).\n')
    steps.add_run('2. Paso 2: ').bold = True
    steps.add_run('Conectan los enfoques con sus hallazgos. Ejemplo:\n')
    steps.add_run('   - "Usaremos ABJ con semillas de plátano para resolver la confusión en conteo".\n')
    steps.add_run('3. Paso 3: ').bold = True
    steps.add_run('Diseñan una ').bold = False
    steps.add_run('actividad flash ').italic = True
    steps.add_run('(10 min) usando esos enfoques. Ejemplo:\n')
    steps.add_run('   - Juego "Pesca numérica": Con cañas de bambú, pescan tarjetas con cantidades representadas de 3 formas (dibujos, números, palabras).')
    
    # Regla
    regla = doc.add_paragraph()
    regla.add_run('Regla: ').bold = True
    regla.add_run('¡Deben usar al menos un material local! (ej: hojas de plátano, tapas recicladas).')
    regla.runs[0].font.color.rgb = RGBColor(255, 0, 0)  # Rojo
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Dinámica 3
    dyn3 = doc.add_heading('Dinámica 3: "Armando la secuencia mágica"', level=2)
    dyn3.add_run(' (40 min)').bold = False
    
    # Actividad
    act3 = doc.add_paragraph()
    act3.add_run('Actividad: ').bold = True
    act3.add_run('Rompecabezas didáctico').italic = True
    
    # Plantilla guía
    plantilla = doc.add_paragraph()
    plantilla.add_run('1. Plantilla guía: ').bold = True
    plantilla.add_run('Cada grupo recibe un esquema de secuencia didáctica con espacios vacíos:\n')
    plantilla.add_run('```\n')
    plantilla.add_run('SESIÓN 1: [Enfoque 1] → Actividad concreta: ___________\n')
    plantilla.add_run('SESIÓN 2: [Enfoque 2] → Representación gráfica: _______\n')
    plantilla.add_run('SESIÓN 3: [Enfoque 3] → Evaluación lúdica: __________\n')
    plantilla.add_run('```\n')
    plantilla.runs[2].font.name = 'Courier New'  # Fuente monoespaciada para el código
    plantilla.runs[3].font.name = 'Courier New'
    plantilla.runs[4].font.name = 'Courier New'
    plantilla.runs[5].font.name = 'Courier New'
    
    # Tiempo para crear
    tiempo = doc.add_paragraph()
    tiempo.add_run('2. Tiempo para crear: ').bold = True
    tiempo.add_run('20 min. Ejemplo real:\n')
    tiempo.add_run('   - Sesión 1 (ABJ): "Carrera de canastos" (contar semillas en equipos).\n')
    tiempo.add_run('   - Sesión 2 (Sociocultural): Graficar ventas del mercado caucasiano.\n')
    tiempo.add_run('   - Sesión 3 (Duval): Transformar cantidades de frutas a símbolos.\n')
    
    # Presentación rápida
    presentacion = doc.add_paragraph()
    presentacion.add_run('3. Presentación rápida: ').bold = True
    presentacion.add_run('Cada grupo expone en 2 minutos con mímica o canción.')
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Dinámica 4
    dyn4 = doc.add_heading('Dinámica 4: "Evaluación con emojis"', level=2)
    dyn4.add_run(' (10 min)').bold = False
    
    # Contenido
    content = doc.add_paragraph()
    content.add_run('- Cada practicante pega un post-it con:\n')
    content.add_run('  - ').bold = True
    content.add_run('1 emoji ').italic = True
    content.add_run('que represente su aprendizaje (ej: 🚀 = "Despegamos con ideas nuevas").\n')
    content.add_run('  - ').bold = True
    content.add_run('1 pregunta ').italic = True
    content.add_run('para la sistematización (ej: "¿Cómo registrar el impacto en los niños?").')
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Productos del taller
    productos = doc.add_heading('Productos del taller:', level=2)
    
    prod_list = [
        "Secuencias didácticas listas para implementar en el Liceo Caucasia.",
        "Banco de ideas: Actividas adaptables a otros grados.",
        "Ruta de sistematización: Pautas para documentar evidencias (fotos, diarios, voces de niños, el diario pedagógico)."
    ]
    
    for prod in prod_list:
        p = doc.add_paragraph(prod, style='List Bullet')
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Claves para los practicantes
    claves = doc.add_heading('Claves para los practicantes:', level=2)
    
    claves_list = [
        "Contexto es todo: Usen materiales caucasianos (ej: piedras del Río Cauca, fibra de plátano).",
        "Flexibilidad: Si una actividad no funciona, ¡improvisen como los arrieros!",
        "Sistematización fácil:\n   - Antes: Video de 1 minuto con expectativas.\n   - Durante: Fotografíen los 'momentos wow' (ej: un niño explicando un concepto).\n   - Después: Entrevista grupal a niños: '¿Qué aprendiste jugando?'."
    ]
    
    for clave in claves_list:
        p = doc.add_paragraph(clave, style='List Bullet')
    
    # Frase final
    frase_final = doc.add_paragraph()
    frase_final.add_run('¡Y recuerden! ').bold = True
    frase_final.add_run('Como dijo un niño de Caucasia: ')
    frase_final.add_run('"Las matemáticas son como el río: si te metes despacio, le agarras el gusto"').italic = True
    frase_final.add_run('.')
    
    # Línea divisoria
    doc.add_paragraph('―'*50).alignment = WD_PARAGRAPH_ALIGNMENT.CENTER
    
    # Nota final
    nota = doc.add_paragraph()
    nota.add_run('Nota: ').bold = True
    nota.add_run('Este taller asegura que las secuencias nazcan de la realidad observada, integren teoría y práctica, y sean ')
    nota.add_run('divertidas para niños y docentes').bold = True
    nota.add_run('. ¡Manos a la obra!')
    
    # Guardar documento
    doc.save('Taller_Secuencias_Caucasia.docx')
    print("Documento creado exitosamente: 'Taller_Secuencias_Caucasia.docx'")

if __name__ == "__main__":
    create_workshop_document()
````
