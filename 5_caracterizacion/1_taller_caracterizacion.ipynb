````python
import pandas as pd
import numpy as np

# Cargar la lista de estudiantes desde el archivo CSV
path = 'C:/Users/marco/Documentos/docencia/groups_list/infantil.xlsx'
estudiantes = pd.read_excel(path)
estudiantes.index = range(1, len(estudiantes) + 1)
estudiantes.head(3) 
````

````python
# Convertir la columna 'Nombre' a una lista de Python
lista_estudiantes = estudiantes['Nombre'].tolist()

# Función para llamar a un estudiante aleatoriamente y registrar su nota de seguimiento
def asistencia():
    global lista_estudiantes
    if lista_estudiantes:
        estudiante_aleatorio = np.random.choice(lista_estudiantes)
        print(f"Llamando a {estudiante_aleatorio}...")
        asistencia = input(f"¿Asistió {estudiante_aleatorio}? (s/n): ").strip().lower()
        
        # Validación de entrada
        if asistencia == 'n':
            # Obtener el índice del estudiante en el DataFrame
            indice = estudiantes[estudiantes['Nombre'] == estudiante_aleatorio].index[0]
            # Incrementar su valor en la columna 'Parcial'
            estudiantes.at[indice, 'Asistencia'] += 4
            print(f"""Actualizado: {estudiante_aleatorio} 
                  ahora tiene {estudiantes.at[indice, 'Asistencia']} en la columna 'Asistencia'.""")
        
        elif asistencia != 's':
            print("Entrada inválida. Por favor, introduce 's' para sí o 'n' para no.")
            return  # Volver a preguntar para el mismo estudiante si la entrada es inválida
        
        # Remover al estudiante de la lista para no llamarlo de nuevo
        lista_estudiantes.remove(estudiante_aleatorio)
    else:
        print("Se han llamado a todos los estudiantes del grupo completo.")

# Función para guardar cambios al final de la sesión
def guardar_cambios():
    estudiantes.to_csv(path, index=False)
    print("Cambios guardados en el archivo CSV.")

# Ejemplo de uso
while lista_estudiantes:
    asistencia()

# Guardar los cambios una vez que todos los estudiantes hayan sido llamados
guardar_cambios()

````

El enfoque STEAMS en educación preescolar, primaria y secundaria es una extensión del enfoque STEM (Ciencia, Tecnología, Ingeniería y Matemáticas) y su variante STEAM, que incorpora el Arte. 

El acrónimo STEAMS incluye:



1. **Science (Ciencia):** Fomenta la curiosidad por el mundo natural y el desarrollo de habilidades científicas, como la observación, experimentación y análisis.



2. **Technology (Tecnología):** Integra herramientas tecnológicas para potenciar el aprendizaje y la resolución de problemas, promoviendo la alfabetización digital desde edades tempranas.



3. **Engineering (Ingeniería):** Introduce el pensamiento lógico y la resolución de problemas estructurados, fomentando habilidades como el diseño y la construcción de soluciones.



4. **Arts (Arte):** Añade creatividad y pensamiento crítico, permitiendo a los estudiantes expresarse y explorar el aprendizaje a través de formas artísticas (plásticas, música, teatro).



5. **Mathematics (Matemáticas):** Refuerza las habilidades numéricas y el razonamiento lógico, fundamentales en la resolución de problemas y comprensión de conceptos abstractos.



6. **Social Sciences (Ciencias Sociales):** Incorpora la comprensión del entorno social, político, cultural y económico, promoviendo el desarrollo de la conciencia crítica sobre el mundo y su historia, además de habilidades de empatía y ciudadanía.



El enfoque STEAMS pretende desarrollar un aprendizaje **interdisciplinario** que integre estas áreas para fomentar un pensamiento más holístico y práctico. 

Se busca que los estudiantes no solo aprendan conocimientos técnicos, sino también cómo aplicarlos creativa y críticamente en situaciones reales, promoviendo habilidades como el trabajo en equipo, la comunicación y la innovación. 

En educación preescolar, primaria, secundaria y superior, se adapta a las diferentes etapas de desarrollo cognitivo y emocional de los estudiantes, promoviendo el aprendizaje activo y contextualizado.

### **Metodología de Enseñanza: Aprendizaje Basado en Problemas y Proyectos (ABPP)**



El **Aprendizaje Basado en Problemas y Proyectos (ABPP)** es una metodología educativa que integra los enfoques del **Aprendizaje Basado en Problemas (ABP)** y el **Aprendizaje Basado en Proyectos (PBL, por sus siglas en inglés)**. 

Es especialmente útil para la formación de futuros docentes, ingenieros y administradores, ya que combina el desarrollo de habilidades prácticas con el pensamiento crítico y reflexivo. 

A continuación, se presenta una descripción adaptada a la enseñanza de las matemáticas en educación infantil, Licenciatura en Matemáticas, Ingeniería Agropecuaria y Administración de Empresas:

---



### **1. Fundamentos del ABPP**


El ABPP se basa en:
- **Resolución de Problemas Reales:** Los estudiantes analizan y resuelven problemas auténticos relacionados con su área de formación, fomentando la aplicación práctica del conocimiento.


- **Proyectos como Herramientas Integradoras:** Los proyectos permiten desarrollar habilidades interdisciplinarias y ofrecer soluciones concretas a los problemas planteados.


- **Aprendizaje Activo:** Los estudiantes son responsables de su aprendizaje y participan activamente en el proceso de investigación, diseño y solución.

---



### **2. Aplicación del ABPP en Áreas Específicas**

#### **Educación Infantil (Licenciatura en Educación Infantil)**
- **Enfoque:** Desarrollo del pensamiento matemático en la infancia desde contextos lúdicos, familiares y culturales.


- **Problemas típicos:** 
  - ¿Cómo enseñar los números naturales mediante actividades prácticas y juegos?
  - ¿Qué estrategias pueden emplearse para abordar el concepto de cantidad en edades tempranas?


- **Proyectos:** Diseño de secuencias didácticas que usen Python, Jupyter Notebooks y materiales manipulativos para enseñar matemáticas de forma interactiva.


- **Competencias desarrolladas:** 
  - Uso de herramientas tecnológicas para diseñar actividades educativas.
  - Comprensión del desarrollo cognitivo infantil y su relación con las matemáticas.



#### **Licenciatura en Matemáticas**
- **Enfoque:** Resolución de problemas matemáticos aplicados y profundización teórica en áreas como análisis matemático y modelado.


- **Problemas típicos:**
  - ¿Cómo se pueden resolver ecuaciones diferenciales aplicadas a modelos físicos o económicos?
  - ¿Cómo modelar fenómenos naturales mediante funciones matemáticas?


- **Proyectos:** 
  - Desarrollo de software educativo para explorar conceptos matemáticos.
  - Aplicación de modelado matemático para resolver problemas reales en otras disciplinas.


- **Competencias desarrolladas:**
  - Dominio de la matemática teórica y aplicada.
  - Uso de tecnología para la enseñanza y la investigación.



#### **Ingeniería Agropecuaria**
- **Enfoque:** Modelado matemático y análisis de datos para mejorar la producción agrícola y la gestión de recursos.
- **Problemas típicos:**
  - ¿Cómo modelar el crecimiento de cultivos considerando variables climáticas?
  - ¿Cómo optimizar el uso del agua en sistemas de riego mediante análisis matemático?


- **Proyectos:** 
  - Desarrollo de modelos predictivos con Python y bibliotecas de aprendizaje automático.
  - Implementación de proyectos que combinen matemáticas y ciencias de datos para resolver problemas agrícolas.


- **Competencias desarrolladas:**
  - Integración de matemáticas con tecnologías avanzadas.
  - Resolución de problemas agropecuarios mediante análisis y modelado.



#### **Administración de Empresas**


- **Enfoque:** Uso de las matemáticas y la ciencia de datos para la toma de decisiones estratégicas.
- **Problemas típicos:**
  - ¿Cómo identificar patrones de comportamiento del consumidor usando series de tiempo?
  - ¿Qué estrategias matemáticas pueden optimizar la asignación de recursos en una empresa?


- **Proyectos:** 
  - Implementación de modelos de análisis financiero y simulaciones.
  - Desarrollo de dashboards interactivos para la gestión de datos empresariales.


- **Competencias desarrolladas:**
  - Habilidad para interpretar datos matemáticos en contextos de negocio.
  - Capacidad para diseñar soluciones basadas en datos.





### **3. Beneficios del ABPP**


- **Desarrollo de competencias clave:**
  - Pensamiento crítico y creativo.
  - Trabajo colaborativo.
  - Comunicación efectiva.
  - Uso de herramientas tecnológicas avanzadas.


- **Adaptación al contexto laboral:** Los estudiantes aprenden a enfrentar desafíos del mundo real y a diseñar soluciones prácticas y aplicables.
- **Flexibilidad e interdisciplinariedad:** La metodología se adapta a diversas áreas, fomentando el aprendizaje integrado.

---



### **4. Ejemplo General de Implementación**


1. **Identificación del problema:** Se plantea un desafío concreto, como modelar el crecimiento poblacional de una comunidad rural (Ingeniería Agropecuaria) o diseñar una estrategia para enseñar fracciones a niños de 8 años (Educación Infantil).


2. **Formación de equipos:** Los estudiantes trabajan en grupos, distribuyendo roles y tareas.
3. **Investigación y diseño:** Los estudiantes investigan el problema, proponen soluciones y diseñan un proyecto.


4. **Implementación:** Se desarrolla el proyecto, como una herramienta tecnológica o una propuesta pedagógica.
5. **Evaluación:** Los estudiantes presentan sus resultados en un informe o una exposición, recibiendo retroalimentación de sus compañeros y docentes.

---



El ABPP es una metodología versátil que conecta el conocimiento teórico con la práctica, preparando a los estudiantes para los retos de sus profesiones en la enseñanza de la matemática, la ingeniería y la administración.

# Diferentes Metodologias activas para la clase de Formación de maestros de Licenciatura en Educación Infantil 



Aquí tienes **5 metodologías innovadoras** para realizar tu taller de formación de maestros, combinando trabajo en parejas, grupos y aprendizaje activo sobre caracterización del contexto y diseño de prácticas pedagógicas para el pensamiento matemático:

---



### **1. Método "Jigsaw" (Rompecabezas)**  
**Objetivo:** Construir colectivamente una guía de caracterización del contexto.  
**Actividades:**  
- **Parejas:** Cada pareja investiga un aspecto de la caracterización (ej: perfil socioeconómico, intereses infantiles, recursos del aula).  
- **Grupos (4-5 personas):** Reunirse para integrar la información en un mapa conceptual.  
- **Producto final:** Infografía colaborativa con estrategias para vincular el contexto al diseño de actividades matemáticas.  

**Herramienta:** Mural.co o póster físico con post-its.  

---



### **2. World Café**  
**Objetivo:** Reflexionar sobre cómo adaptar prácticas pedagógicas a distintos contextos.  
**Actividades:**  
- **Mesas temáticas (grupos de 4):** Cada mesa discute una pregunta (ej: *"¿Cómo identificar las ideas matemáticas informales de los niños en su juego libre?"*).  
- **Rotación:** Cada 15 minutos, un integrante queda como "anfitrión" y los demás rotan para enriquecer las respuestas.  
- **Parejas:** Al final, cada pareja diseña una actividad matemática basada en las conclusiones.  

**Material:** Hojas grandes para registrar ideas en cada mesa.  

---



### **3. Aprendizaje Basado en Retos (ABR)**  
**Objetivo:** Crear herramientas de diagnóstico contextual.  
**Actividades:**  
- **Reto inicial (grupos de 3):** *"Diseñen un instrumento para caracterizar las habilidades matemáticas de niños en un aula rural multigrado"*.  
- **Parejas:** Prototipado rápido (ej: entrevista lúdica, diario de observación).  
- **Presentación:** Cada grupo expone su herramienta y recibe feedback con la técnica *"2 estrellas y 1 deseo"* (2 aspectos positivos + 1 sugerencia).  

**Ejemplo de producto:** Rúbrica de observación con indicadores de pensamiento matemático.  

---



### **4. Role-Playing + Simulación**  
**Objetivo:** Practicar la observación y adaptación pedagógica.  
**Actividades:**  
- **Parejas:** Uno actúa como maestro que aplica una actividad matemática (ej: clasificación de objetos), el otro como niño con necesidades específicas (ej: dificultad en lenguaje).  
- **Grupos (6 personas):** Discuten cómo modificar la actividad según el perfil del "niño".  
- **Plenaria:** Socialización de ajustes pedagógicos.  

**Recurso:** Videos cortos de niños reales en actividades matemáticas (para análisis grupal).  

---



### **5. Método "Design Thinking"**  
**Objetivo:** Diseñar prácticas pedagógicas centradas en el niño.  
**Fases:**  
1. **Empatizar (parejas):** Entrevistas simuladas para identificar necesidades de los niños.  
2. **Definir (grupos):** Sintetizar hallazgos (ej: *"Los niños aprenden mejor con materiales concretos de su entorno"*).  
3. **Prototipar (grupos):** Crear una actividad matemática usando solo recursos locales (ej: semillas, piedras).  
4. **Testear (parejas):** Un maestro implementa la actividad, el otro observa y registra.  

**Producto:** Kit de actividades matemáticas contextualizadas.  

---



### **Elementos Clave para Todas las Metodologías**  
- **Documentación:** Portafolio digital con fotos, videos y reflexiones.  
- **Técnicas de cierre:**  
  - *"El semáforo"*: Verde (lo aprendido), amarillo (dudas), rojo (dificultades).  
  - *"Tweet ficticio"*: Resumir en 140 caracteres una idea clave.  

**Duración sugerida:** 4 horas (2 sesiones de 2h con breaks activos).  

---



### **Ejemplo de Actividad Concreta**  
**Nombre:** *"El Mapa Matemático de Nuestra Escuela"*  
- **Parejas:** Identifican espacios físicos de la escuela donde emergen conceptos matemáticos (ej: escaleras → secuencias numéricas).  
- **Grupos:** Diseñan una secuencia didáctica usando esos espacios.  
- **Producto:** Mapa interactivo con códigos QR que vinculan lugares a actividades matemáticas.  



---

Estas metodologías promueven **aprendizaje colaborativo, reflexión crítica y aplicación práctica**, esenciales para que los maestros vinculen la caracterización del contexto con el diseño de prácticas pedagógicas efectivas.
