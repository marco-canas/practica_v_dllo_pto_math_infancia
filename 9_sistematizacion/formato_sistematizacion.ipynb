### **1. Fase de Observación y Caracterización**  
**Objetivo:** Documentar el contexto educativo y las necesidades de los estudiantes.  

| **Aspecto**               | **Descripción**                                                                 | **Evidencias** (ej: notas, fotos, entrevistas) | **Reflexión** (¿Qué aprendí?) |
|---------------------------|---------------------------------------------------------------------------------|-----------------------------------------------|------------------------------|
| Contexto socioeducativo   | Características de la institución, estudiantes, familias y entorno.             |                                               |                              |
| Necesidades identificadas | Problemas de aprendizaje, recursos disponibles, desafíos pedagógicos.          |                                               |                              |
| Perfil del grupo          | Estilos de aprendizaje, intereses, dinámicas grupales.                         |                                               |                              |

**Preguntas guía:**  
- ¿Cómo influye el contexto en los procesos de aprendizaje?  
- ¿Qué necesidades prioritarias surgieron?  

---

### **2. Fase de Selección de Enfoques Pedagógicos**  
**Objetivo:** Fundamentar teóricamente los enfoques elegidos y su pertinencia.  

| **Enfoque seleccionado** | **Justificación teórica** (¿Por qué este enfoque?) | **Relación con el contexto** | **Limitaciones** |
|--------------------------|----------------------------------------------------|-----------------------------|------------------|
| Ej: Constructivismo      |                                                    |                             |                  |
| Ej: Aprendizaje basado en proyectos |                                            |                             |                  |
| Ej: Pedagogía crítica    |                                                    |                             |                  |

**Preguntas guía:**  
- ¿Los enfoques seleccionados responden a las necesidades identificadas?  
- ¿Qué autores o teorías sustentan esta elección?  



### **3. Hipótesis Investigativa**  
**Planteamiento inicial:**  
*"Los tres enfoques pedagógicos escogidos realmente resuelven las necesidades educativas de los estudiantes del Liceo Caucasia."*  

| **Variable**           | **Indicadores de verificación** (¿Cómo sabré si se cumple?) | **Resultados esperados** |
|------------------------|------------------------------------------------------------|--------------------------|
| Aplicación de enfoques | Estrategias implementadas, participación de estudiantes.   |                          |
| Impacto en aprendizaje | Mejoras en comprensión, motivación, habilidades.           |                          |

**Reflexión:**  
- ¿La hipótesis fue ajustada durante el proceso? ¿Por qué?  



### **4. Diseño e Implementación de Clases**  
**Plantilla por sesión (5 clases):**  

| **Clase #** | **Objetivo de aprendizaje** | **Enfoque aplicado** | **Actividades** (descripción breve) | **Resultados observados** | **Dificultades y ajustes** |
|-------------|-----------------------------|----------------------|-------------------------------------|---------------------------|----------------------------|
| 1           |                             |                      |                                     |                           |                            |
| 2           |                             |                      |                                     |                           |                            |

**Preguntas guía:**  
- ¿Los objetivos de cada clase contribuyeron a resolver el problema inicial?  
- ¿Qué recursos o estrategias fueron más efectivos?  



### **5. Evaluación y Conclusiones**  
**Matriz de logros y desafíos:**  

| **Aspecto evaluado**       | **Logros**                                          | **Dificultades**                              | **Propuestas de mejora** |
|----------------------------|-----------------------------------------------------|-----------------------------------------------|--------------------------|
| Planeación de clases       |                                                     |                                               |                          |
| Implementación de enfoques |                                                     |                                               |                          |
| Impacto en los estudiantes |                                                     |                                               |                          |

**Conclusiones finales:**  
1. ¿Se validó la hipótesis inicial? ¿Por qué?  
2. ¿Qué aprendizajes clave obtuve de esta práctica?  
3. Recomendaciones para futuras intervenciones.  



### **Anexos**  
- **Evidencias:** Fotografías, planeaciones de clase, trabajos de estudiantes.  
- **Bitácora de reflexión:** Registro diario de observaciones y insights.  
- **Instrumentos de evaluación:** Rúbricas, encuestas aplicadas.  



### **Instrucciones para su uso:**  
1. **Documentación continua:** Completa cada sección durante el proceso (no al final).  
2. **Enfoque reflexivo:** Usa las preguntas guía para profundizar en el análisis.  
3. **Evidencias concretas:** Adjunta archivos o referencias que sustenten tus hallazgos.  



```python
from docx import Document
from docx.shared import Pt
from docx.enum.text import WD_ALIGN_PARAGRAPH

def crear_documento_pedagogico():
    # Crear un nuevo documento
    doc = Document()
    
    # Configuración de estilo base
    style = doc.styles['Normal']
    font = style.font
    font.name = 'Arial'
    font.size = Pt(11)

    # --- Título principal ---
    titulo = doc.add_heading('SISTEMATIZACIÓN DE LA PRÁCTICA PEDAGÓGICA', level=1)
    titulo.alignment = WD_ALIGN_PARAGRAPH.CENTER
    
    # --- Información inicial ---
    doc.add_paragraph('Liceo Caucasia')
    doc.add_paragraph('Practicante 1: ___________________________')
    doc.add_paragraph('Practicante 2: ___________________________')
    doc.add_paragraph('Grado asignado: _______       Período: ________')
    doc.add_paragraph()  # Espacio

    # --- 1. Fase de Observación ---
    doc.add_heading('1. Fase de Observación y Caracterización', level=2)
    doc.add_paragraph('Objetivo: Documentar el contexto educativo y las necesidades de los estudiantes.', style='Intense Quote')

    # Tabla de observación
    table_obs = doc.add_table(rows=4, cols=4)
    table_obs.style = 'Table Grid'
    
    # Encabezados
    hdr_cells = table_obs.rows[0].cells
    hdr_cells[0].text = 'Aspecto'
    hdr_cells[1].text = 'Descripción'
    hdr_cells[2].text = 'Evidencias (ej: notas, fotos, entrevistas)'
    hdr_cells[3].text = 'Reflexión (¿Qué aprendí?)'

    # Filas de contenido
    aspectos = [
        ("Contexto socioeducativo", "Características de la institución, estudiantes, familias y entorno."),
        ("Necesidades identificadas", "Problemas de aprendizaje, recursos disponibles, desafíos pedagógicos."),
        ("Perfil del grupo", "Estilos de aprendizaje, intereses, dinámicas grupales.")
    ]
    
    for i, (aspecto, desc) in enumerate(aspectos, start=1):
        row_cells = table_obs.rows[i].cells
        row_cells[0].text = aspecto
        row_cells[1].text = desc
        row_cells[2].text = ""
        row_cells[3].text = ""

    # Preguntas guía
    doc.add_paragraph('Preguntas guía:', style='Heading 3')
    doc.add_paragraph('- ¿Cómo influye el contexto en los procesos de aprendizaje?')
    doc.add_paragraph('- ¿Qué necesidades prioritarias surgieron?')
    doc.add_paragraph('---')

    # --- 2. Fase de Enfoques Pedagógicos ---
    doc.add_heading('2. Fase de Selección de Enfoques Pedagógicos', level=2)
    doc.add_paragraph('Objetivo: Fundamentar teóricamente los enfoques elegidos y su pertinencia.', style='Intense Quote')

    # Tabla de enfoques
    table_enfoques = doc.add_table(rows=4, cols=4)
    table_enfoques.style = 'Table Grid'
    
    # Encabezados
    hdr_cells = table_enfoques.rows[0].cells
    hdr_cells[0].text = 'Enfoque seleccionado'
    hdr_cells[1].text = 'Justificación teórica (¿Por qué este enfoque?)'
    hdr_cells[2].text = 'Relación con el contexto'
    hdr_cells[3].text = 'Limitaciones'

    # Ejemplos pre-llenados
    ejemplos = [
        ("Constructivismo", "", "", ""),
        ("Aprendizaje basado en proyectos", "", "", ""),
        ("Pedagogía crítica", "", "", "")
    ]
    
    for i, (enfoque, *_) in enumerate(ejemplos, start=1):
        row_cells = table_enfoques.rows[i].cells
        row_cells[0].text = enfoque

    # Preguntas guía
    doc.add_paragraph('Preguntas guía:', style='Heading 3')
    doc.add_paragraph('- ¿Los enfoques seleccionados responden a las necesidades identificadas?')
    doc.add_paragraph('- ¿Qué autores o teorías sustentan esta elección?')
    doc.add_paragraph('---')

    # --- 3. Hipótesis Investigativa ---
    doc.add_heading('3. Hipótesis Investigativa', level=2)
    doc.add_paragraph('Planteamiento inicial:', style='Heading 3')
    doc.add_paragraph('"Los tres enfoques pedagógicos escogidos realmente resuelven las necesidades educativas de los estudiantes del Liceo Caucasia."', style='Intense Quote')

    # Tabla de hipótesis
    table_hipotesis = doc.add_table(rows=3, cols=3)
    table_hipotesis.style = 'Table Grid'
    
    # Encabezados
    hdr_cells = table_hipotesis.rows[0].cells
    hdr_cells[0].text = 'Variable'
    hdr_cells[1].text = 'Indicadores de verificación (¿Cómo sabré si se cumple?)'
    hdr_cells[2].text = 'Resultados esperados'

    # Filas de contenido
    variables = [
        ("Aplicación de enfoques", "Estrategias implementadas, participación de estudiantes."),
        ("Impacto en aprendizaje", "Mejoras en comprensión, motivación, habilidades.")
    ]
    
    for i, (var, ind) in enumerate(variables, start=1):
        row_cells = table_hipotesis.rows[i].cells
        row_cells[0].text = var
        row_cells[1].text = ind

    # Reflexión
    doc.add_paragraph('Reflexión:', style='Heading 3')
    doc.add_paragraph('- ¿La hipótesis fue ajustada durante el proceso? ¿Por qué?')
    doc.add_paragraph('---')

    # --- 4. Diseño de Clases ---
    doc.add_heading('4. Diseño e Implementación de Clases', level=2)
    doc.add_paragraph('Plantilla por sesión (5 clases):', style='Heading 3')

    # Tabla de clases
    table_clases = doc.add_table(rows=6, cols=6)
    table_clases.style = 'Table Grid'
    
    # Encabezados
    hdr_cells = table_clases.rows[0].cells
    hdr_cells[0].text = 'Clase #'
    hdr_cells[1].text = 'Objetivo de aprendizaje'
    hdr_cells[2].text = 'Enfoque aplicado'
    hdr_cells[3].text = 'Actividades (descripción breve)'
    hdr_cells[4].text = 'Resultados observados'
    hdr_cells[5].text = 'Dificultades y ajustes'

    # Filas vacías para 5 clases
    for i in range(1, 6):
        row_cells = table_clases.rows[i].cells
        row_cells[0].text = str(i)

    # Preguntas guía
    doc.add_paragraph('Preguntas guía:', style='Heading 3')
    doc.add_paragraph('- ¿Los objetivos de cada clase contribuyeron a resolver el problema inicial?')
    doc.add_paragraph('- ¿Qué recursos o estrategias fueron más efectivos?')
    doc.add_paragraph('---')

    # --- 5. Evaluación y Conclusiones ---
    doc.add_heading('5. Evaluación y Conclusiones', level=2)

    # Matriz de logros
    doc.add_paragraph('Matriz de logros y desafíos:', style='Heading 3')
    table_eval = doc.add_table(rows=4, cols=4)
    table_eval.style = 'Table Grid'
    
    # Encabezados
    hdr_cells = table_eval.rows[0].cells
    hdr_cells[0].text = 'Aspecto evaluado'
    hdr_cells[1].text = 'Logros'
    hdr_cells[2].text = 'Dificultades'
    hdr_cells[3].text = 'Propuestas de mejora'

    # Filas de contenido
    aspectos_eval = [
        "Planeación de clases",
        "Implementación de enfoques",
        "Impacto en los estudiantes"
    ]
    
    for i, aspecto in enumerate(aspectos_eval, start=1):
        row_cells = table_eval.rows[i].cells
        row_cells[0].text = aspecto

    # Conclusiones finales
    doc.add_paragraph('Conclusiones finales:', style='Heading 3')
    doc.add_paragraph('1. ¿Se validó la hipótesis inicial? ¿Por qué?')
    doc.add_paragraph('2. ¿Qué aprendizajes clave obtuve de esta práctica?')
    doc.add_paragraph('3. Recomendaciones para futuras intervenciones.')
    doc.add_paragraph('---')

    # --- Anexos ---
    doc.add_heading('Anexos', level=2)
    doc.add_paragraph('- Evidencias: Fotografías, planeaciones de clase, trabajos de estudiantes.')
    doc.add_paragraph('- Bitácora de reflexión: Registro diario de observaciones y insights.')
    doc.add_paragraph('- Instrumentos de evaluación: Rúbricas, encuestas aplicadas.')
    doc.add_paragraph('---')

    # --- Instrucciones ---
    doc.add_heading('Instrucciones para su uso:', level=2)
    doc.add_paragraph('1. Documentación continua: Completa cada sección durante el proceso (no al final).')
    doc.add_paragraph('2. Enfoque reflexivo: Usa las preguntas guía para profundizar en el análisis.')
    doc.add_paragraph('3. Evidencias concretas: Adjunta archivos o referencias que sustenten tus hallazgos.')

    # Guardar documento
    doc.save('Sistematizacion_Practica_Pedagogica.docx')
    print("Documento generado con éxito: 'Sistematizacion_Practica_Pedagogica.docx'")

# Ejecutar la función
crear_documento_pedagogico()
```
