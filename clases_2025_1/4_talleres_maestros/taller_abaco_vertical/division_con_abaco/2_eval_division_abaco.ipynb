### 4. Mapa conceptual  
El mapa conceptual se generará utilizando Graphviz en Python. Aquí está el código para crearlo:  

```python
from graphviz import Digraph

# Crear el gráfico
dot = Digraph(comment='Clase de división con ábaco vertical', format='png')
dot.attr('node', shape='box', style='filled', color='lightblue')
dot.attr('edge', color='gray')

# Nodos
dot.node('A', 'División con ábaco vertical')
dot.node('B', 'Partes del ábaco')
dot.node('C', 'Algoritmo de división')
dot.node('D', 'Simulación en Python')
dot.node('E', 'Historia del ábaco')
dot.node('F', 'Arte y diseño')
dot.node('G', 'Trabajo en equipo')

# Edges
dot.edges(['AB', 'AC', 'AD', 'AE', 'AF', 'AG'])

# Renderizar y guardar
dot.render('C:\\Users\\marco\\Descargas\\mapa_conceptual', view=True)
```
