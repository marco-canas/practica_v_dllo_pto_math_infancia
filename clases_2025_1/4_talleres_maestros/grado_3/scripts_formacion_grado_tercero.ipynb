```python
import pandas as pd

# Leer el archivo como texto plano
with open('DBA_G3_MATEMATICAS.dba', 'r', encoding='utf-8') as f:
    lineas = f.readlines()

# Procesamiento de secciones
secciones = []
current_section = {}
in_section = False

for linea in lineas:
    linea = linea.strip()
    
    # Detectar inicio de sección
    if linea.startswith(':ENUNCIADO'):
        current_section = {
            'tipo': 'enunciado',
            'contenido': []
        }
        in_section = True
    elif linea.startswith(':EVIDENCIAS'):
        current_section = {
            'tipo': 'evidencias',
            'contenido': []
        }
        in_section = True
    elif linea.startswith(':EJEMPLO'):
        current_section = {
            'tipo': 'ejemplo',
            'contenido': []
        }
        in_section = True
    
    # Capturar contenido de sección
    elif in_section:
        if linea.startswith('ECHO'):
            contenido = linea.replace('ECHO ', '').strip('" ')
            if contenido not in ['', '------------------------------------------------------------------']:
                current_section['contenido'].append(contenido)
        else:
            in_section = False
            if current_section['contenido']:
                secciones.append(current_section)

# Crear DataFrame estructurado
df = pd.DataFrame(secciones)
df['seccion'] = df.groupby(df.index // 3).ngroup() + 1  # Agrupar en tripletes
df = df.pivot(index='seccion', columns='tipo', values='contenido').applymap(lambda x: '\n'.join(x))

# Mostrar resultado
df
```

```python
# Ejemplo: Obtener el primer enunciado
print(df.loc[4, 'enunciado'])
```

```python

# Ejemplo: Listar todas las evidencias
for idx, row in df.iterrows():
    print(f"Evidencias sección {idx}:")
    print(row['evidencias'])
    print("\n")
```
