```python


````python
import os

# Definir la ruta base
base_path = r"C:\Users\marco\Documentos\docencia\practica_v_dllo_pto_math_infancia\clases_2025_1\3_enfoques_pedagogicos"

# Crear las 10 carpetas
for i in range(1, 11):
    folder_name = f"Enfoque_{i}"
    folder_path = os.path.join(base_path, folder_name)
    
    # Crear la carpeta si no existe
    os.makedirs(folder_path, exist_ok=True)
    print(f"Carpeta creada: {folder_path}")

print("\n¡Todas las carpetas han sido creadas exitosamente!")

````


Explicación del script:

1. `import os`: Importa el módulo necesario para trabajar con el sistema operativo

2. `base_path`: La ruta donde se crearán las carpetas, usando un string raw (`r""`) para manejar correctamente las barras invertidas en Windows

3. El bucle `for` itera de 1 a 10 (el `range(1, 11)` genera números del 1 al 10)

4. `folder_name`: Genera el nombre de cada carpeta usando f-strings

5. `folder_path`: Combina la ruta base con el nombre de cada carpeta

6. `os.makedirs()`: Crea el directorio con dos parámetros importantes:
   - `folder_path`: Ruta completa a crear
   - `exist_ok=True`: Evita errores si la carpeta ya existe

7. Los mensajes `print()` muestran el progreso y confirmación final

Ejecuta este script y se crearán las carpetas con la estructura solicitada. Si alguna carpeta ya existe, el script simplemente la ignorará gracias al parámetro `exist_ok=True`.

# Scripts para la gestión de los archivos .md para la presentación de cada uno de los 11 enfoques pedagógicos para práctica V

1. **Aprendizaje Basado en Retos (ABR)**
   - Divide la clase en grupos y asigna a cada grupo un enfoque pedagógico.
   - Cada equipo debe diseñar y demostrar una actividad didáctica basada en su enfoque.
   - Los demás compañeros hacen preguntas y reflexionan sobre su aplicabilidad.

2. **Taller de Experiencias Sensoriales**
   - Prepara estaciones con materiales concretos (semillas, bloques, hojas) que representen distintos enfoques.
   - Los estudiantes rotan entre estaciones y registran cómo cada enfoque favorece el pensamiento matemático.

3. **Role-Playing Docente**
   - Cada estudiante asume el papel de un maestro y expone un enfoque pedagógico a sus "alumnos" (sus compañeros).
   - Debe incluir una mini actividad didáctica.
   - La clase brinda retroalimentación sobre claridad y aplicabilidad.

4. **Infografías y Visual Thinking**
   - Los estudiantes crean infografías digitales o murales visuales explicando un enfoque pedagógico.
   - Se exponen en el aula o en redes sociales académicas (ej. Padlet, Instagram académico).

5. **Círculo Socrático con Preguntas Críticas**
   - Se plantean preguntas como:
     - ¿Qué enfoque es más viable en un aula rural sin tecnología?
     - ¿Cómo se combinan varios enfoques en una misma actividad?
   - Los estudiantes debaten en círculo, justificando con ejemplos.

6. **Escape Room Pedagógico**
   - Se diseñan acertijos y desafíos donde cada pista lleva a descubrir un enfoque pedagógico.
   - Los estudiantes deben relacionar teorías con prácticas didácticas para avanzar.

7. **Creación de Recursos Didácticos**
   - Cada grupo diseña un recurso didáctico (video, cuento, canción) aplicando un enfoque.
   - Se presentan en clase y se suben a una plataforma colaborativa.

### 🖥 **Script en Python para Crear un Archivo CSV con los 11 Enfoques**
Este código generará un archivo `enfoques_pedagogicos.csv` con el nombre, descripción y justificación de cada enfoque.


````python
import csv

# Lista de enfoques pedagógicos
enfoques = [
    ["Aprendizaje Basado en Juegos (ABJ)", "Integrar juegos estructurados para desarrollar conceptos matemáticos.", "Promueve el aprendizaje lúdico y el desarrollo sensoriomotor."],
    ["Enfoque Sociocultural con Pertinencia Local", "Vincular las matemáticas a prácticas culturales y económicas de la región.", "Alinea con los lineamientos curriculares del MEN."],
    ["Resolución de Problemas Cotidianos", "Plantear problemas reales del entorno para el desarrollo de habilidades matemáticas.", "Desarrolla competencias para 'saber hacer' en contextos reales."],
    ["Manipulación de Material Concreto", "Uso de objetos tangibles para enseñar operaciones básicas y geometría.", "Favorece la etapa sensoriomotora y operacional concreta en niños."],
    ["Proyectos de Aula con Enfoque Investigativo", "Diseñar secuencias didácticas basadas en preguntas de los niños.", "Fortalece el eje investigativo y la sistematización de datos."],
    ["Modelación Matemática con Contextos Rurales", "Representar matemáticamente situaciones del campo y la naturaleza.", "Alinea con los Derechos Básicos de Aprendizaje (DBA)."],
    ["Pensamiento Visible a través del Arte", "Uso de dibujos, música y danza para expresar conceptos matemáticos.", "Fomenta la expresión creativa y la integración del arte con la matemática."],
    ["Aprendizaje Cooperativo con Roles Comunitarios", "Trabajo en grupos con roles inspirados en la vida cotidiana.", "Fomenta la comunicación matemática y la interdependencia positiva."],
    ["Enfoque Reggio Emilia Adaptado", "Espacios de exploración autónoma con elementos naturales.", "Empodera a los niños como constructores activos del conocimiento."],
    ["Integración de TIC con Recursos Accesibles", "Uso de apps y herramientas digitales para reforzar conceptos matemáticos.", "Desarrolla competencias digitales sin perder la reflexión crítica."],
    ["Teoría de los Registros de Representación Semiótica (Raymond Duval)", "Transformar y coordinar diferentes representaciones matemáticas.", "Facilita la comprensión abstracta mediante múltiples registros."]
]

# Crear archivo CSV
with open("enfoques_pedagogicos.csv", mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Enfoque", "Descripción", "Justificación"])  # Escribir encabezados
    writer.writerows(enfoques)

print("Archivo CSV generado con éxito.")

````
