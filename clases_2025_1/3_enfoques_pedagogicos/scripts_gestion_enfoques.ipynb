````python
import os

# Definir la ruta base
base_path = r"C:\Users\marco\Documentos\docencia\practica_v_dllo_pto_math_infancia\clases_2025_1\3_enfoques_pedagogicos"

# Crear las 10 carpetas
for i in range(1, 11):
    folder_name = f"Enfoque_{i}"
    folder_path = os.path.join(base_path, folder_name)
    
    # Crear la carpeta si no existe
    os.makedirs(folder_path, exist_ok=True)
    print(f"Carpeta creada: {folder_path}")

print("\n¡Todas las carpetas han sido creadas exitosamente!")

````


Explicación del script:

1. `import os`: Importa el módulo necesario para trabajar con el sistema operativo

2. `base_path`: La ruta donde se crearán las carpetas, usando un string raw (`r""`) para manejar correctamente las barras invertidas en Windows

3. El bucle `for` itera de 1 a 10 (el `range(1, 11)` genera números del 1 al 10)

4. `folder_name`: Genera el nombre de cada carpeta usando f-strings

5. `folder_path`: Combina la ruta base con el nombre de cada carpeta

6. `os.makedirs()`: Crea el directorio con dos parámetros importantes:
   - `folder_path`: Ruta completa a crear
   - `exist_ok=True`: Evita errores si la carpeta ya existe

7. Los mensajes `print()` muestran el progreso y confirmación final

Ejecuta este script y se crearán las carpetas con la estructura solicitada. Si alguna carpeta ya existe, el script simplemente la ignorará gracias al parámetro `exist_ok=True`.
