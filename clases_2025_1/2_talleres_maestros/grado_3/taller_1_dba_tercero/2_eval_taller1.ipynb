# 6. Actividades de Aprendizaje  
## **Juego 1: Mercado Interactivo**  
```python
import ipywidgets as widgets

def mercado_interactivo(frutas, precio_kg):
    total = np.sum(frutas * precio_kg)
    plt.bar(['Manzanas', 'Peras', 'Uvas'], frutas, color=['red', 'green', 'purple'])
    plt.title(f'Total: ${total}')
    plt.show()

widgets.interact(
    mercado_interactivo,
    frutas=widgets.IntSlider(value=[3, 4, 2], min=0, max=10),
    precio_kg=widgets.Fixed([1500, 2000, 3000])
)
```

## **Animación: Carrera de Operaciones**  
```python
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

fig, ax = plt.subplots()
x = np.arange(0, 4*np.pi, 0.1)
line, = ax.plot(x, np.sin(x))

def animate(i):
    line.set_ydata(np.sin(x + i/10))
    ax.set_title(f'Función Seno: Iteración {i}')
    return line,

ani = FuncAnimation(fig, animate, frames=100, interval=50)
HTML(ani.to_jshtml())
```

# 8. Tareas de Ejercitación  
## **Problema con SymPy**  
```python
from sympy import symbols, Eq, solve

manzanas, peras = symbols('m p')
sistema = [
    Eq(3*manzanas + 2*peras, 15000),
    Eq(manzanas - peras, 1000)
]
sol = solve(sistema, (manzanas, peras))
print(f"$$\\text{{Solución: }} m = {sol[manzanas]}, p = {sol[peras]}$$")
```
